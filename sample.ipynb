```python
from PIL import Image
import os
os.environ["PATH"] += os.pathsep + r"C:\Program Files\Tesseract-OCR"
import pytesseract
import pandas as pd
import re

```

```python
player_stats_raw = pytesseract.image_to_string(Image.open('data/Unedited.jpg'))
# Parse into key-value pairs
data = {}
for line in player_stats_raw.splitlines():
    line = line.strip()
    if not line:
        continue

    # Match "Label: Value" pattern
    match = re.match(r'^(.+?):\s*([\d,]+)$', line)
    if match:
        key = match.group(1).strip()
        value = int(match.group(2).replace(',', ''))  # Remove commas, convert to int
        data[key] = value

df = pd.DataFrame(list(data.items()), columns=['Metric', 'Value'])
print(df)
```
